```python
import json
import os
import datetime
```

```python
jsonFilesPath = "../1_FnoStock"
```

```python
jsonFiles = [d for d in os.listdir(jsonFilesPath) if ".json" in d and "data.json" not in d]
dates = ["21-02-2025", "24-02-2025", "25-02-2025", "27-02-2025", "28-02-2025", "03-03-2025", "04-03-2025", "05-03-2025", "06-03-2025"]
```

```python
MainDic = {}
for i in range(len(jsonFiles)):
    # Open and read a JSON file
    with open(jsonFilesPath+"/"+jsonFiles[i], "r") as file:
        data = json.load(file)
    date = dates[i]
    
    MainDic[date] = {}
    for key in data.keys():
        tmpDic = {}
        script = key
        PriceCng = data[key]['PriceVol']['PriceCng']
        FutOiPer = data[key]['FutureData']['FutOiPer']
        Fut_category = data[key]['FutureData']['Fut_category']
        
        MainDic[date][script] = {"PriceCng": PriceCng, "FutOiPer":FutOiPer, "Fut_category" : Fut_category}
```

```python
# Write JSON data to a file
with open("FutOiHistory.json", "w") as file:
    json.dump(MainDic, file, indent=4)  # `indent=4` makes it readable
```

```python
with open("data.json", "r") as FutOihistory:
    data = json.load(FutOihistory)
```
